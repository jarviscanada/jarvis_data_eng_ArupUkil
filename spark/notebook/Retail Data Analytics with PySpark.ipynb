# Retail Data Analytics with PySpark

Re-implementation of
`python_data_analytics/python_data_wrangling/retail_data_analytics_wrangling.ipynb`
for Databricks using PySpark DataFrames.

This notebook keeps the original business flow:
- ingest retail data from PostgreSQL or CSV
- clean and standardize the schema
- compute invoice, monthly order, sales, active user, and RFM metrics
- use `display(...)` for Databricks-native result inspection

```python
from pyspark.sql import functions as F
from pyspark.sql import Window
```

## Clean and Standardize the Schema

The original CSV uses mixed naming such as `Customer ID`. Here we rename columns,
cast data types, and add consistent string cleanup.

```python
retail_df = (
    raw_retail_df
    .withColumnRenamed("Invoice", "invoice_no")
    .withColumnRenamed("StockCode", "stock_code")
    .withColumnRenamed("Description", "description")
    .withColumnRenamed("Quantity", "quantity")
    .withColumnRenamed("InvoiceDate", "invoice_date")
    .withColumnRenamed("Price", "unit_price")
    .withColumnRenamed("Customer ID", "customer_id")
    .withColumnRenamed("Country", "country")
    .withColumn("invoice_no", F.trim(F.col("invoice_no").cast("string")))
    .withColumn("stock_code", F.trim(F.col("stock_code").cast("string")))
    .withColumn("description", F.trim(F.coalesce(F.col("description").cast("string"), F.lit(""))))
    .withColumn("quantity", F.col("quantity").cast("int"))
    .withColumn("invoice_date", F.to_timestamp("invoice_date"))
    .withColumn("unit_price", F.col("unit_price").cast("double"))
    .withColumn("customer_id", F.col("customer_id").cast("double"))
    .withColumn("country", F.trim(F.coalesce(F.col("country").cast("string"), F.lit(""))))
)

display(retail_df)
retail_df.printSchema()
```

```python
retail_df.write.mode("overwrite").saveAsTable(retail_table_name)
display(spark.table(retail_table_name))
```

## Prepare the Analytics DataFrame

```python
analysis_df = (
    retail_df
    .withColumn("line_amount", F.col("quantity") * F.col("unit_price"))
    .withColumn("is_canceled", F.col("invoice_no").startswith("C"))
    .withColumn("year_month", F.date_format("invoice_date", "yyyy-MM"))
    .withColumn(
        "year_month_int",
        (F.year("invoice_date") * F.lit(100)) + F.month("invoice_date"),
    )
    .cache()
)

display(analysis_df)
```

## Total Invoice Amount Distribution

```python
invoice_amount_df = (
    analysis_df
    .groupBy("invoice_no")
    .agg(F.round(F.sum("line_amount"), 2).alias("invoice_amount"))
)

positive_invoice_amount_df = invoice_amount_df.filter(F.col("invoice_amount") > 0)

invoice_mode_df = (
    positive_invoice_amount_df
    .groupBy("invoice_amount")
    .count()
    .orderBy(F.desc("count"), F.asc("invoice_amount"))
    .limit(1)
    .select(F.col("invoice_amount").alias("mode_invoice_amount"))
)

invoice_amount_summary_df = (
    positive_invoice_amount_df
    .agg(
        F.round(F.min("invoice_amount"), 2).alias("minimum_invoice_amount"),
        F.round(F.avg("invoice_amount"), 2).alias("mean_invoice_amount"),
        F.round(F.expr("percentile_approx(invoice_amount, 0.5)"), 2).alias("median_invoice_amount"),
        F.round(F.max("invoice_amount"), 2).alias("maximum_invoice_amount"),
        F.round(F.expr("percentile_approx(invoice_amount, 0.85)"), 2).alias("p85_invoice_amount"),
    )
    .crossJoin(invoice_mode_df)
)

display(invoice_amount_summary_df)
display(positive_invoice_amount_df.orderBy(F.desc("invoice_amount")))
```

```python
p85_invoice_amount = invoice_amount_summary_df.first()["p85_invoice_amount"]

invoice_amount_first_85_df = positive_invoice_amount_df.filter(
    F.col("invoice_amount") <= F.lit(p85_invoice_amount)
)

display(invoice_amount_first_85_df.orderBy(F.desc("invoice_amount")))
```

## Monthly Placed and Canceled Orders

```python
order_df = (
    analysis_df
    .select("invoice_no", "invoice_date", "year_month", "year_month_int", "is_canceled")
    .dropDuplicates(["invoice_no"])
)

monthly_orders_df = (
    order_df
    .groupBy("year_month", "year_month_int")
    .agg(
        F.countDistinct("invoice_no").alias("total_orders"),
        F.sum(F.col("is_canceled").cast("int")).alias("canceled_orders"),
    )
    .withColumn("placed_orders", F.col("total_orders") - (F.lit(2) * F.col("canceled_orders")))
    .withColumn(
        "cancellation_rate_pct",
        F.round(
            (F.col("canceled_orders") / (F.col("placed_orders") + F.col("canceled_orders"))) * F.lit(100),
            2,
        ),
    )
    .orderBy("year_month_int")
)

display(monthly_orders_df)
```

## Monthly Sales

```python
valid_sales_df = (
    analysis_df
    .filter(
        (~F.col("is_canceled")) &
        (F.col("quantity") > 0) &
        (F.col("unit_price") > 0)
    )
    .cache()
)

monthly_sales_df = (
    valid_sales_df
    .groupBy("year_month", "year_month_int")
    .agg(F.round(F.sum("line_amount"), 2).alias("sales_amount"))
    .orderBy("year_month_int")
)

display(monthly_sales_df)
```

## Monthly Sales Growth

```python
month_window = Window.orderBy("year_month_int")

monthly_sales_growth_df = (
    monthly_sales_df
    .withColumn("previous_sales_amount", F.lag("sales_amount").over(month_window))
    .withColumn(
        "pct_growth",
        F.round(
            ((F.col("sales_amount") - F.col("previous_sales_amount")) / F.col("previous_sales_amount")) * 100,
            2,
        ),
    )
    .drop("previous_sales_amount")
)

display(monthly_sales_growth_df)
```

## Monthly Active Users

```python
valid_customer_sales_df = valid_sales_df.filter(F.col("customer_id").isNotNull()).cache()

monthly_active_users_df = (
    valid_customer_sales_df
    .groupBy("year_month", "year_month_int")
    .agg(F.countDistinct("customer_id").alias("active_users"))
    .orderBy("year_month_int")
)

display(monthly_active_users_df)
```

## New and Existing Users

```python
first_purchase_month_df = (
    valid_customer_sales_df
    .groupBy("customer_id")
    .agg(F.min("year_month_int").alias("first_purchase_month_int"))
)

user_months_df = (
    valid_customer_sales_df
    .select("customer_id", "year_month", "year_month_int")
    .dropDuplicates()
    .join(first_purchase_month_df, on="customer_id", how="left")
    .withColumn(
        "user_type",
        F.when(F.col("year_month_int") == F.col("first_purchase_month_int"), F.lit("new"))
        .otherwise(F.lit("existing")),
    )
)

monthly_user_types_df = (
    user_months_df
    .groupBy("year_month", "year_month_int")
    .pivot("user_type", ["new", "existing"])
    .agg(F.count(F.lit(1)))
    .na.fill(0)
    .orderBy("year_month_int")
)

display(monthly_user_types_df)
```

## RFM Metrics

RFM stands for:
- Recency
- Frequency
- Monetary value

```python
rfm_base_df = (
    analysis_df
    .filter(F.col("customer_id").isNotNull())
    .groupBy("customer_id")
    .agg(
        F.datediff(F.to_date(F.lit(snapshot_date)), F.max(F.to_date("invoice_date"))).alias("Recency"),
        F.countDistinct("invoice_no").alias("Frequency"),
        F.round(F.sum("line_amount"), 2).alias("Monetary"),
    )
)

recency_window = Window.orderBy(F.col("Recency").asc(), F.col("customer_id").asc())
frequency_window = Window.orderBy(F.col("Frequency").asc(), F.col("customer_id").asc())
monetary_window = Window.orderBy(F.col("Monetary").asc(), F.col("customer_id").asc())

rfm_df = (
    rfm_base_df
    .withColumn("RecencyScore", (F.lit(6) - F.ntile(5).over(recency_window)).cast("string"))
    .withColumn("FrequencyScore", F.ntile(5).over(frequency_window).cast("string"))
    .withColumn("MonetaryScore", F.ntile(5).over(monetary_window).cast("string"))
    .withColumn("RFM_SCORE", F.concat("RecencyScore", "FrequencyScore", "MonetaryScore"))
)

display(rfm_df.orderBy(F.desc("Monetary")))
display(rfm_df.select("Recency", "Frequency", "Monetary").summary())
```

## RFM Segmentation

```python
rfm_segmented_df = (
    rfm_df
    .withColumn("rf_score", F.concat("RecencyScore", "FrequencyScore"))
    .withColumn(
        "Segment",
        F.when(F.col("rf_score").rlike(r"[1-2][1-2]"), "hibernating")
        .when(F.col("rf_score").rlike(r"[1-2][3-4]"), "at_risk")
        .when(F.col("rf_score").rlike(r"[1-2]5"), "cant_lose")
        .when(F.col("rf_score").rlike(r"3[1-2]"), "about_to_sleep")
        .when(F.col("rf_score").rlike(r"33"), "need_attention")
        .when(F.col("rf_score").rlike(r"[3-4][4-5]"), "loyal_customers")
        .when(F.col("rf_score").rlike(r"41"), "promising")
        .when(F.col("rf_score").rlike(r"51"), "new_customers")
        .when(F.col("rf_score").rlike(r"[4-5][2-3]"), "potential_loyalists")
        .when(F.col("rf_score").rlike(r"5[4-5]"), "champions")
        .otherwise("other")
    )
)

segment_summary_df = (
    rfm_segmented_df
    .groupBy("Segment")
    .agg(
        F.count("*").alias("customers"),
        F.round(F.sum("Monetary"), 2).alias("total_revenue"),
        F.round(F.avg("Recency"), 2).alias("avg_recency"),
        F.round(F.avg("Frequency"), 2).alias("avg_frequency"),
        F.round(F.avg("Monetary"), 2).alias("avg_monetary"),
    )
)

total_revenue_value = segment_summary_df.agg(F.sum("total_revenue").alias("total_revenue")).first()["total_revenue"]

segment_summary_df = (
    segment_summary_df
    .withColumn(
        "revenue_pct",
        F.round((F.col("total_revenue") / F.lit(total_revenue_value)) * F.lit(100), 2),
    )
    .orderBy(F.desc("customers"))
)

display(segment_summary_df)
```
